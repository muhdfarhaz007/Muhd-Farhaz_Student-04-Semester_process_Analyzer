## 4. NumPy practice

```python
# Select the three semester marks columns.
marks = df[["Semester_1", "Semester_2", "Semester_3"]]

# Convert the Pandas table into a NumPy array.
marks_array = marks.to_numpy()
print("NumPy array:")
print(marks_array)

# Calculate basic statistics with NumPy.
print("\nAverage of all marks:", np.mean(marks_array))
print("Highest mark:", np.max(marks_array))
print("Lowest mark:", np.min(marks_array))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
analysis_df = df.copy()

# Calculate the average for each student (axis=1 means across columns).
analysis_df["Average"] = analysis_df[
    ["Semester_1", "Semester_2", "Semester_3"]
].mean(axis=1).round(2)

# Compare Semester 3 marks with Semester 1 marks.
analysis_df["Change_S1_to_S3"] = (
    analysis_df["Semester_3"] - analysis_df["Semester_1"]
)

# Filter students whose marks improved.
improving_students = analysis_df[analysis_df["Change_S1_to_S3"] > 0]
print("Students who improved:")
display(improving_students[["Name", "Semester_1", "Semester_3", "Change_S1_to_S3"]])

# Sort from highest average to lowest average.
ranked_students = analysis_df.sort_values(
    by="Average", ascending=False
).reset_index(drop=True)
ranked_students.insert(0, "Rank", range(1, len(ranked_students) + 1))
display(ranked_students)
```

## 6. Your individual project challenge
**Challenge:** Calculate average; identify students improving or declining from Semester 1 to Semester 3; rank by average.

Write your solution below and explain what the result means.


```python
# STEP 1: Calculate every student's average across the three semesters.
result = df.copy()
semester_columns = ["Semester_1", "Semester_2", "Semester_3"]
result["Average"] = result[semester_columns].mean(axis=1).round(2)

# STEP 2: Find the change from Semester 1 to Semester 3.
# Positive = improved, negative = declined, zero = no change.
result["Change_S1_to_S3"] = result["Semester_3"] - result["Semester_1"]

# STEP 3: Convert the change into an easy-to-read progress label.
def check_progress(change):
    if change > 0:
        return "Improving"
    elif change < 0:
        return "Declining"
    else:
        return "No Change"

result["Progress"] = result["Change_S1_to_S3"].apply(check_progress)

# STEP 4: Rank students. The highest average receives Rank 1.
result = result.sort_values(
    by="Average", ascending=False
).reset_index(drop=True)
result.insert(0, "Rank", range(1, len(result) + 1))

print("FINAL STUDENT PROGRESS REPORT")
display(result)

# STEP 5: Print useful findings.
print("Improving students:", (result["Progress"] == "Improving").sum())
print("Declining students:", (result["Progress"] == "Declining").sum())
print("No change:", (result["Progress"] == "No Change").sum())

print("\nHighest-ranked student:")
display(result.head(1)[["Rank", "Name", "Average", "Change_S1_to_S3", "Progress"]])

print("Lowest-ranked student:")
display(result.tail(1)[["Rank", "Name", "Average", "Change_S1_to_S3", "Progress"]])

print("\nAverage mark in each semester:")
display(df[semester_columns].mean().round(2))
```
